```python
import pandas as pd
import numpy as np
```

Read data from different sources and combine then into one samples.tsv

```python
samples = pd.read_stata("resources/Baza_zadanie_051026.dta", index_col='KYH_ID')
samples = samples.set_index(np.int64(samples.index))
samples_old = pd.read_stata("resources/260505_Alex_Zobolev.dta", index_col="KYH_ID")
runs = pd.read_csv("resources/SraRunTable.csv", index_col = "Sample Name")
ids_map = pd.read_csv("resources/ids_map.csv")
```

```python
print('runs, ids_map:', len(runs.index.symmetric_difference(ids_map['new_id'])))
print('samples, ids_map:', len(samples.index) - sum(ids_map['ID_KYH'].isin(samples.index)))
```

```python
meta = (runs.reset_index()
            .merge(ids_map, left_on='Sample Name', right_on='new_id')
            .merge(samples, left_on='ID_KYH', right_index=True))
meta['batch'] = 'B1'
```

```python
meta[[
    'Assay Type',
    'AvgSpotLen',
    'Center Name',
    'Collection_Date',
    'Instrument',
    'lat_lon',
    'LibraryLayout',
    'LibrarySelection',
    'LibrarySource',
    'Platform',
    'ReleaseDate',
    'create_date',
    'version'
]].apply(lambda c: (len(c.unique()), c.unique()), result_type='expand').T
```
